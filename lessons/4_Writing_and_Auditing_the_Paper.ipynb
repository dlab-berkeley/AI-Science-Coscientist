# Rules for Using AI Co-Scientists

- **The evaluator is the science.** Evolution only improves what you measure.
- **Check whether it discovered or remembered.** Disguise the data when you can.
- **Use the uncertainty that matches the data.** Resample the independent unit.
- **Verify every citation and number.** Open the source; rerun the code.
- **Keep unpublished data off free tiers.** Free plans may train on what you paste.
- **Disclose AI use.** You own every claim in the paper.

💡 **Tip:** Across the workshop, debate ranked a wrong hypothesis first, the
held-out score found the right one, and the fluent write-up still needed
checking line by line.

## ❗ Key Points

* A template paper reads its numbers from the analysis, so it cannot misquote them.
* A research agent writes a fluent paper fast, but fluency is not accuracy.
* Audit every claim of an AI-written paper against the results and the sources.
* Our agent-written paper had six errors, including the headline result.
* You own every claim: verify, disclose AI use, and keep unpublished data off free tiers.
