## 🥊 Challenge 1: A Leaky Evaluator

```python
%%writefile memorizer.py
import math


def fit(train_rows):
    def predict(x1, x2):
        nearest = min(train_rows, key=lambda row: abs(math.log(row['x1']) - math.log(x1)))
        return nearest['y']

    return predict
```

## 🥊 Challenge 2: Write a Better Candidate

```python
%%writefile my_candidate.py
import math

import numpy as np


def fit(train_rows):
    design = np.array([[1, math.log10(row['x1']), math.log10(row['x2'])] for row in train_rows])
    target = np.array([math.log10(row['y']) for row in train_rows])
    (log_c, p, q), *_ = np.linalg.lstsq(design, target, rcond=None)

    def predict(x1, x2):
        return 10 ** log_c * x1 ** p * x2 ** q

    return predict
```
